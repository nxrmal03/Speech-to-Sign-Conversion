```python
import os

def transcribe_audio_folder(folder_path):
    audio_files = [f for f in os.listdir(folder_path) if f.endswith('.mp3')]
    transcriptions = {}
    for audio_file in audio_files:
        try:
            result = model.transcribe(os.path.join(folder_path, audio_file),language="tamil")
            transcriptions[audio_file] = result['text']
        except Exception as e:
            transcriptions[audio_file] = f"Error: {str(e)}"
            
    return transcriptions

folder_path = "TAMIL MP3"
transcriptions = transcribe_audio_folder(folder_path)
for audio_file, transcription in transcriptions.items():
    print(f"Audio File: {audio_file}")
    print(f"Transcription: {transcription}")
```
